```python
# Setup: player stats (run this first)
player_name = "Alex"
player_hp = 30
player_gold = 0
weapon = "rusty sword"

print("Welcome to the Lighthouse Depths, " + player_name + "!")
print("HP: " + str(player_hp) + " | Gold: " + str(player_gold) + " | Weapon: " + weapon)
```

```python
#Combat round (fixed number of turns with range)
enemy = "Cave Goblin"
enemy_hp = 12
player_attack = 4
enemy_attack = 3

print("A " + enemy + " jumps out of the shadows!")
for turn in range(1, 4):
    print("--- Turn " + str(turn) + " ---")
    enemy_hp = enemy_hp - player_attack
    print("You swing your " + weapon + ". The " + enemy + " has " + str(enemy_hp) + " HP left.")
    if enemy_hp > 0:
        player_hp = player_hp - enemy_attack
        print("The " + enemy + " scratches you! You have " + str(player_hp) + " HP left.")

if enemy_hp <= 0:
    print("The " + enemy + " collapses. Victory!")
else:
    print("The " + enemy + " runs off into the dark.")
```

```python
#Loot (loop over chests, running gold total)
print("You find a row of 5 old chests.")
for chest in range(1, 6):
    if chest == 3:
        print("Chest " + str(chest) + ": empty... except for a spider.")
    elif chest == 5:
        gold_found = 50
        print("Chest " + str(chest) + ": a golden goblet! +" + str(gold_found) + " gold")
        player_gold = player_gold + gold_found
    else:
        gold_found = chest * 10
        print("Chest " + str(chest) + ": +" + str(gold_found) + " gold")
        player_gold = player_gold + gold_found
    print("   Running total: " + str(player_gold) + " gold")

print("Total gold: " + str(player_gold))
```

```python
#Countdown (negative step)
print("Your torch is burning out...")
for flame in range(5, 0, -1):
    print("Torch: " + "*" * flame + " (" + str(flame) + ")")
print("Darkness.")
print()

print("Water is pouring into the cave!")
for depth in range(10, -1, -2):
    if depth == 0:
        print("The water reaches your chin. SWIM!")
    else:
        print(str(depth) + " feet until the water reaches you...")
```

```python
#Map (nested loops, end=" " for rows)
rows = 5
cols = 7
player_row = 2
player_col = 1
exit_row = 3
exit_col = 5

for r in range(rows):
    for c in range(cols):
        if r == player_row and c == player_col:
            print("P", end=" ")
        elif r == exit_row and c == exit_col:
            print("E", end=" ")
        elif r == 0 or r == rows - 1 or c == 0 or c == cols - 1:
            print("#", end=" ")
        else:
            print(".", end=" ")
    print()  # move to the next row

print("P = you   E = exit   # = wall")
```

```python
#Riddle door (palindrome check with reversed slice)
print("A stone door with a carved face speaks:")
print('"Say a word that reads the same forwards and backwards."')

guess = input("Your answer: ")
guess = guess.lower()
backwards = guess[::-1]
print("Backwards that is: " + backwards)

if len(guess) > 2 and guess == backwards:
    print("The door grinds open. Well spoken, " + player_name + ".")
else:
    print("The face frowns. The door stays shut.")
```

```python
#Cursed scroll (decode with slicing or replace)
line1 = "ecalperif eht dniheb si pam ehT"
line2 = "Txhxex xkxexyx xixsx xuxnxdxexrx xtxhxex xbxexlxlx"
line3 = "Trust#the#lighthouse#keeper"

print("The scroll is gibberish... until you look closer.")
print("Line 1: " + line1[::-1])                # written backwards
print("Line 2: " + line2[::2])                 # every 2nd letter is junk
print("Line 3: " + line3.replace("#", " "))    # # marks are spaces
```

```python
#Player title from name initial (indexing)
initial = player_name[0].upper()

if initial in "AEIOU":
    title = "the Arcane"
elif initial in "BCDFGH":
    title = "the Brave"
elif initial in "JKLMNP":
    title = "the Clever"
elif initial in "QRSTV":
    title = "the Swift"
else:
    title = "the Shadow"

last = player_name[-1].lower()
print("Your name begins with " + initial + " and ends with " + last + ".")
print("From now on you are known as: " + player_name + " " + title)
print("Short title: " + player_name[:3].upper() + "-" + title[4:].upper())
```

```python
#Room events (find/count on descriptions)
room = "A damp cave. Gold coins glint near a gold statue. A loose stone looks like a trap."
print(room)
print()

gold_count = room.lower().count("gold")
print("You spot " + str(gold_count) + " things made of gold.")

trap_spot = room.find("trap")
if trap_spot != -1:
    print("WARNING: You notice a trap! (found at position " + str(trap_spot) + ")")
else:
    print("The room looks safe.")

if room.find("monster") == -1:
    print("No monsters here... for now.")
```

```python
# Week 5: The captain's log (found in the shipwreck room)
log = "ENTRY-0427:Fog rolling in near the eastern rocks"

# Split the label from the message using find + slices
colon = log.find(":")
label = log[:colon]
message = log[colon + 1:]

# Pull the entry number out of the label
dash = label.find("-")
number = label[dash + 1:]
print("Entry number: " + number)
print("Message: " + message)

# Make a short preview of the message
preview = message[:12] + "..."
print("Preview: " + preview)

# Censor a word
safe = message.replace("rocks", "*****")
print("Censored: " + safe)

# Count the vowels with a loop
vowels = 0
for i in range(len(message)):
    if message[i] in "aeiouAEIOU":
        vowels = vowels + 1
print("Vowels: " + str(vowels))

# Game hooks: the log gives the player clues
if int(number) > 400:
    print("The ink on this entry is still wet... someone wrote it recently.")
if message.find("eastern") != -1:
    print("Clue unlocked: head EAST.")
```
